## Benchmarks

* Perplexity
    * Perplexity is the method to assess performance of the LLMs
    * Lower the value, better the performance 

* GLUE Benchmark
    * grammatical correctness (CoLA) and sentiment polarity (SST-2) of individual sentences.
    * Similarity, and Paraphrase Tasks, focuses on assessing the model's capacity to identify paraphrases in sentence pairs
    *  Inference Tasks, challenges the model to handle sentence entailment and relationships.
* SuperGLUE Benchmark
    * Boolean question answering, textual entailment, coreference resolution, reading comprehension with commonsense reasoning, and word sense disambiguation.
* BIG-Bench Benchmark
    * 204 or more language tasks that cover a wide range of topics and languages
    * is considered a "living benchmark," accepting new task submissions for continuous peer review
* HELM Benchmark
* FLASK Benchmark

```python
# sample perplexity code 
import numpy as np

def perplexity(probabilities):
    sentence_probablity = probabilities.prod()
    sentence_probablity_norm = sentence_probablity ** (1/len(probabilities))
    perplexity = 1 / (sentence_probablity_norm)
    return perplexity


probabilities = np.array([0.4, 0.27, 0.55,0.79])
res = perplexity(probabilities)
print(res)

# Suppose we further train the LLM, and the probabilities of the next best word become higher

probabilities = np.array([0.2, 0.4, 0.6, 0.8])
res = perplexity(probabilities)
print(res)
```

```python
import deeplake, os
from dotenv import load_dotenv

load_dotenv()


def create_deeplake_dataset(dataset):
    return deeplake.dataset(f"hub://{os.getenv('ACTIVELOOP_ORG_ID')}/{dataset}")

    
ds = create_deeplake_dataset('test_dataset')
```

```python
# add some texts to the dataset
ds.create_tensor('text', htype="text")

texts = [f"text {i}" for i in range(1, 11)]
for text in texts:
    ds.append({"text": text})
```

```python
batch_size=3

train_loader = ds.dataloader() \
.batch(batch_size)\
.shuffle()\
.pytorch()


for i, batch in enumerate(train_loader):
    print(f' batch{i}')
    samples = batch.get('text')
    for j, sample in enumerate(samples):
        print(f'sample {j}: {sample}')
        
    print()
    pass
    
```

```python
from torch.utils.data import DataLoader, Dataset

class DeepLakePyTorchDataset(Dataset):
    def __init__(self, ds):
        self.ds = ds

    def __len__(self):
        return len(self.ds)

    def __getitem__(self, idx):
        texts = self.ds.text[idx].text().astype(str)
        return { "text": texts }
    
```

```python
ds_pt = DeepLakePyTorchDataset(ds)

dl_pt = DeepLakePyTorchDataset(ds_pt, batch_size=3, shuffle=True)
```
