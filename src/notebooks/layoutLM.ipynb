```python
# Use a pipeline as a high-level helper
from paddleocr import PaddleOCR, draw_ocr
import cv2
from matplotlib import pyplot as plt

# Initialize PaddleOCR model
ocr = PaddleOCR(use_angle_cls=True, lang='en')  # Enable English language and angle classification
```

```python
receipt_image_path = '/Users/rc/Desktop/target.png'

# Load the image using OpenCV
image = cv2.imread(receipt_image_path)
```

```python
import pdfplumber

file_path = "/Users/rc/workspaces/datasets/reciepts/target/Order details - Walmart.com.pdf"

# Step 1: Open the PDF using pdfplumber
with pdfplumber.open(file_path) as pdf:
    # Step 2: Extract tables from the first page
    first_page = pdf.pages[0]
    print(first_page)
    tables = first_page.extract_tables()
    print(tables)

    # Step 3: Process the extracted tables
    for table in tables:
        for row in table:
            # Assume the structure of each row is [Item, Price]
            print(f"Item: {row[0]}, Price: {row[1]}")
```
